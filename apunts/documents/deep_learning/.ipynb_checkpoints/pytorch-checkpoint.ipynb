## Tensors

L'estructura de dades fonamental de Pytorch és el `tensor`. Un tensor és un array de 0 a moltes dimensions amb unes propietats especials. 

Té uns mètodes similars als arrays de `Numpy`:

```python
scalar = torch.tensor(3.14159)
vector = torch.tensor([1, 2, 3])
matrix = torch.ones((2, 3), dtype=torch.float)
tensor = torch.randn((2, 3, 4), dtype=torch.float)
print(scalar)
print(vector)
print(matrix)
print(tensor)
```

Una comprovació que es fà moltes vegades és mirar la forma del tensor. Observem com `shape` mostra les dimensions del tensor i en el cas de l'escalar és també un tensor amb 0 dimensions. 

```python
print(tensor.shape, scalar.shape)
```

Si aquesta forma no ens interessa, es pot canviar amb `view` o `reshape` (amb mutació)

```python
# Obtenemos un tensor con una forma diferente pero sigue siendo
# el MISMO tensor
same_matrix = matrix.view(1, 6)
# Si cambiamos uno de sus elementos...
same_matrix[0, 1] = 2.
# Cambia ambas variables: matrix y same_matrix
print(matrix)
print(same_matrix)

# Sigamos la sugerencia de PyTorch y usemos el método "clone"
another_matrix = matrix.view(1, 6).clone().detach()
# De nuevo, si cambiamos uno de sus elementos...
another_matrix[0, 1] = 4.
# ¡El tensor original (matrix) queda intacto!
print(matrix)
print(another_matrix)
```

Detach elimina el tensor del grafo computacional. 

Si volem transformar altres tipus de dades en tensors utilitzen `.to_tensor()`. Si no volem la doble precisió, podem utilitzar `.float()`. Per a convertir a numpy `.numpy()`. 

```python
x_train_tensor = torch.as_tensor(np.random.randn(2,3))
print(x_train_tensor.dtype)
float_tensor = x_train_tensor.float()
print(float_tensor.dtype)
np_array = x_train_tensor.numpy()
print(np_array)
```

## Exemple de regressió lineal amb Pytorch

Per il·lustrar nous conceptes dels `tensors` anem a resoldre una regressió lineal amb descens de gradient manualment amb Pytorch i les funcions a baix nivell que tenen els tensors. 

Començarem a la creació de unes dades aleatories. Es poden crear en Numpy i passar a tensors o crear directament les dades en tensors. 

```python
true_b = 1
true_w = 2
N = 100

# Data Generation
torch.manual_seed(42)
#Equivalent a np.random.seed(42)

x = torch.randn(N, dtype=torch.float)
epsilon = (.5 * torch.randn(N, dtype=torch.float))
y = true_b + true_w * x + epsilon
```

A continuació anem a fer una operació que no es sol fer, ja que la separació de train i test es sol fer amb `Dataloaders` que els veurem més endavant. Però per poder utilitzar `random_split` convertim x i y en un `TensorDataset` fem la separació. 

> La metodologia per ingerir i separar el dataset correctament es veurà més endavant.

```python
from torch.utils.data import Dataset, TensorDataset, DataLoader 
from torch.utils.data.dataset import random_split

dataset = TensorDataset(x,y)

train_data, val_data = random_split(dataset, [80, 20])

# Extreure els tensors corresponents a train_data utilitzant els índexs del Subset
x_train_tensor = train_data.dataset.tensors[0][train_data.indices]
y_train_tensor = train_data.dataset.tensors[1][train_data.indices]

x_val_tensor = val_data.dataset.tensors[0][val_data.indices]
y_val_tensor = val_data.dataset.tensors[1][val_data.indices]
```

Dibuixem el scatterplot. En aquest cas també hem tingut que convertir a array Numpy. El procés és `detach`, `cpu`, `numpy` i `flatten` per fer l'array unidimensional. Després entendrem el perquè de totes aquestes funcions.  

```python


x_np = x.detach().cpu().numpy().flatten()
y_np = y.detach().cpu().numpy().flatten()

x_train_np = x_train_tensor.detach().cpu().numpy()
y_train_np = y_train_tensor.detach().cpu().numpy()

x_val_np = x_val_tensor.detach().cpu().numpy()
y_val_np = y_val_tensor.detach().cpu().numpy()


sns.scatterplot(x=x_np, y=y_np, alpha=0.7)
sns.scatterplot(x=x_val_np, y=y_val_np, alpha=0.7)
```

Els tensors tenen la capacitat de estar en memòria RAM o en la VRAM de la GPU. Anem a crear els tensors que necessitem per a fer la regressió lineal directament en `GPU` i transformar les dades de train i test a tensors del dispositiu. Per a que funcione aquest notebook en qualsevol ordinador donarem a triar `cuda` o `cpu` 

```python
# Initializes parameters "b" and "w" randomly since we want to apply gradient descent on
# these parameters we need to set REQUIRES_GRAD = TRUE
device = 'cuda' if torch.cuda.is_available() else 'cpu'

torch.manual_seed(42)
b = torch.randn(1, requires_grad=True, dtype=torch.float, device=device)
w = torch.randn(1, requires_grad=True, dtype=torch.float, device=device)
print(b, w)

# Si el dataset fora numpy
#x_train_tensor = torch.as_tensor(x_train).float().to(device)
#y_train_tensor = torch.as_tensor(y_train).float().to(device)

x_train_tensor = x_train_tensor.to(device)
y_train_tensor = y_train_tensor.to(device)
```

Ara tenim inicialitzats els parametres b i w. Entrenarem un model de regressió lineal manualment per veure com es comporten els tensors i les ferramentes de baix nivell que ens dona Pytorch. No és el procediment que es sol seguir. 

A continuació calcularem la `y` resultat d'aplicar els parametres `b` i `w`. Aquesta seguirà una línia que no es pareixerà per a res a la línia que es pot apreciar al gràfic anterior. 

```python
yhat = b + w * x_train_tensor
sns.scatterplot(x=x_np, y=y_np, alpha=0.7)
sns.scatterplot(x=x_train_np, y=yhat.detach().cpu().numpy())
```

Cada punt de la línia de la predicció està a una distància del punt real, això és l'error. El càlcul del `MSE` ens donarà quan equivocat està el model. Triant la funció `MSE` estem triant la funció de perduda. Aquesta pot ser aplicada un a un a tots els punts i seria un descens de gradient `estocastic`. Es pot aplicar a tots els punts i seria per `batches`. Si s'aplica sobre subconjunts estem aplicant un descens de gradient per `mini-batches`. Com que són pocs punts farem un `batch`. 

> Estocàstic suposaria un algorisme lent, per batches pot ser que ocupe tota la memòria si la mida del dataset és molt gran. Per `mini-batches` és una manera d'anar iterant en totes les dades sense omplir la memòria i amb bona velocitat. 

```python
# Computes the loss
# We are using ALL data points, so this is BATCH gradient descent
# How wrong is our model? That's the error! 
error = (yhat - y_train_tensor)
# It is a regression, so it computes mean squared error (MSE)
loss = (error ** 2).mean()
```

```python
print(loss)
print(error)
```

`loss` no és un tensor normal. Aquest guarda la informació dels tensors que tenen `requires_grad=True` que han provocat aquest error. La funció `backward` evita tenir que calcular els gradients manualment. 

Funciona gràcies a **Autograd**. Aquest és un paquet de `diferenciació automàtica`. Evita que tingam que calcular derivades parcials, regla de la cadena i demés. Està per baix del nostre codi dins dels tensors involucrats al càlcul de `loss`. Els tensors tenen un graf que registra quins tensors han participat en la seua obtenció. El mètode de `retropropagació` utilitza el la regla de la cadena, recorre el graf i calcula les derivades parcials.  

```python
# Step 3 - Computes gradients for both "b" and "w" parameters
# No more manual computation of gradients! 
# b_grad = 2 * error.mean()
# w_grad = 2 * (x_tensor * error).mean()
loss.backward()
```

```python
print(b.grad,w.grad, b, w)
```

Ara els tensors b i w tenen un paràmetre anomenat `grad`. això són els valors del gradients. Aquests valors de gradient indiquen la direcció i la magnitud del canvi necessari en els pesos i el biaix per reduir l'error en el següent pas de l'optimització.

Ara tocaria modificar els valors de b i w en funció del seu `grad` i el `learning rate` i tornar a fer el càlcul de `loss` i `backward` fins que l'error siga 0, un màxim que hem definit o fins arrivar al màxim d'iteracions. Però abans de fer-ho cal solucionar un altre problema. 

Si executem de nou el `backward` els gradients s'acumulen. Ho fa així per poder treballar amb `mini-batches`, però nosaltres deguem en cada iteració posar els gradients a zero. 

```python
b.grad.zero_(), w.grad.zero_()
```

Si repetim els passos anteriors cada vegada la b i la w es pareixeran més a les originals i cada vegada la correcció serà menor perquè el gradient detectarà que està més aprop del mínim de la funció de perduda.

Per suposat es pot posar tot en un `for` i muntar un descens de gradient manual. 

```python
# Sets learning rate - this is "eta" ~ the "n"-like Greek letter
lr = 0.1

# Step 0 - Initializes parameters "b" and "w" randomly
torch.manual_seed(42)
b = torch.randn(1, requires_grad=True, \
                dtype=torch.float, device=device)
w = torch.randn(1, requires_grad=True, \
                dtype=torch.float, device=device)

# Defines number of epochs
n_epochs = 1000

for epoch in range(n_epochs):
    # Step 1 - Computes model's predicted output - forward pass
    yhat = b + w * x_train_tensor
    
    # Step 2 - Computes the loss
    # We are using ALL data points, so this is BATCH gradient
    # descent. How wrong is our model? That's the error!
    error = (yhat - y_train_tensor)
    # It is a regression, so it computes mean squared error (MSE)
    loss = (error ** 2).mean()

    # Step 3 - Computes gradients for both "b" and "w" parameters
    # No more manual computation of gradients! 
    # b_grad = 2 * error.mean()
    # w_grad = 2 * (x_tensor * error).mean()   
    # We just tell PyTorch to work its way BACKWARDS 
    # from the specified loss!
    loss.backward()
    
    # Step 4 - Updates parameters using gradients and 
    # the learning rate. But not so fast...
    # FIRST ATTEMPT - just using the same code as before
    # AttributeError: 'NoneType' object has no attribute 'zero_'
    # b = b - lr * b.grad
    # w = w - lr * w.grad
    # print(b)

    # SECOND ATTEMPT - using in-place Python assigment
    # RuntimeError: a leaf Variable that requires grad
    # has been used in an in-place operation.
    # b -= lr * b.grad
    # w -= lr * w.grad        
    
    # THIRD ATTEMPT - NO_GRAD for the win!
    # We need to use NO_GRAD to keep the update out of
    # the gradient computation. Why is that? It boils 
    # down to the DYNAMIC GRAPH that PyTorch uses...
    with torch.no_grad():
        b -= lr * b.grad
        w -= lr * w.grad
    
    # PyTorch is "clingy" to its computed gradients, we
    # need to tell it to let it go...
    b.grad.zero_()
    w.grad.zero_()
    
print(b, w)
```

Aquest codi conté un exemple primitiu de descens de gradient en Pytorch. Com hem fer per separat abans, definim b i w com al pesos o paràmetres del model. en bucle per 1000 `epochs` calculem les prediccions, l'error MSE. Aquest es guarda en una variable `loss` que té la capacitat de saber cóm s'ha fet el seu càlcul. Per això `loss` té la funció `backward` per calcular els gradients de les variables que són `requires_grad=True` que han influit en el seu càlcul. 

La part novedosa és l'ajust de les variables. En una regessió lineal amb descens de gradient amb `Numpy` es fa la resta del gradient calculat a la variable. Ací canvien dos coses:
* El gradient no es calcula amb la funció de cost derivada. Ja ens ve donat al executar `backward` en `loss`.
* L'actualització no es pot fer a les variables tal qual, s'ha de posar en un bloc `with torch.no_grad():` El motiu es veu al següent apartat:

Els quadres blaus representen a b i w. Es veu com la w ha sigut multiplicada per x i després sumada a b. Finalment el resultat és un tensor de 80 elements. No mostra els tensors que no requereixen càlcul de gradients.

Ací tenim el graf de `loss` i es veu com es guardar tot el que ha passat i pot tornar enrerre. 

Amb el bloc with torch.no_grad():, li dius literalment a PyTorch:

"El que faré ara (actualitzar els pesos) és una operació purament mecànica d'àlgebra. No vull que l'afegeixes al mapa de càlcul, no m'iguales gradients d'això, només canvia el valor numèric de la variable."

PyTorch et bloqueja l'actualització directa perquè vol protegir el seu sistema de càlcul de gradients de qualsevol canvi descontrolat que pugui corrompre el graf.

El graf de computació de PyTorch no és etern. Només serveix per a una sola passada cap endavant (forward) i cap enrere (backward). Un cop acabada la iteració i netejats els gradients amb zero_(), aquest graf es descarta.

## Descens de gradient amb Optimitzadors

Fins ara hem actualitzat dos paràmetres amb els gradients manualment. Però si tenim milions d'ells ja no és viable fer-ho manualment. Amés, no sempre és tant fàcil com restar el gradient. De vegades cal aplicar tècniques que eviten l'Overfitting o els mínims locals. Per a això tenim els `optimitzadors` com `SGD`, `RMSprop` o `Adam`. 

![distints optimitzadors](./opt2.gif)

Cada optimitzador té una estratègia distinta per a arribar a al mínim. 

Un optimitzador es crea amb els paràmetres a actualitzar, el `learning rate` i altres hiperparàmetres i s'actualitza amb `step()`. 

Tampoc cal posar els gradients a zero. Això es pot fer en `zero_grad()`. 

El següent exemple utilitza `SGD` que és un descens de gradient comú. En commpte de ser un regressió lineal és una simple funció de pèrduda quadràtica. 

```python
import torch
from IPython.display import display, Math

# parámetro inicial
theta = torch.tensor(3.0, requires_grad=True)  # Este parametro se debe optimiza requires_grad
# optimizador, se encarga de actualizar theta
optimizer = torch.optim.SGD([theta], lr=0.1)  
#### El optimizador le pasas los parametros y el learning rate

# bucle de optimización (30 pasos)
for step in range(30):
    # ponemos a cero los gradientes
    optimizer.zero_grad()  
    # función de pérdida theta^2
    loss = theta**2
    # cálculo de gradientes
    loss.backward()
    # actualización de parámetros
    optimizer.step()
    # mostramos el progreso fancy en latex
    display(Math(
    rf"\text{{Paso }} {step + 1}: \ \theta \text{{ se ha movido a }} {theta.item():.3f},\ "
    rf"\text{{loss}} = {loss.item():.4f}"
    ))
```

A continuació el mateix codi guardant la trajectoria i els grads per fer una animanció

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# ----- 1. RUN GRADIENT DESCENT AND STORE TRAJECTORY -----

theta = torch.tensor(3.0, requires_grad=True)
optimizer = torch.optim.SGD([theta], lr=0.1)

trajectory = []
grads = []

for step in range(30):
    trajectory.append(theta.item())
    optimizer.zero_grad()
    loss = theta**2
    loss.backward()
    grads.append(theta.grad.item())   # store gradient 2θ
    optimizer.step()


trajectory = np.array(trajectory)
grads = np.array(grads)
```

```python
grads
```

```python
# ----- 2. PLOT SETUP -----

fig, ax = plt.subplots(figsize=(6, 4))

# Draw parabola
x = np.linspace(-3, 3, 400)
ax.plot(x, x**2, label="f(θ)=θ²")

# Red point showing θ
point, = ax.plot([], [], 'ro', markersize=8)

# Gradient arrow
arrow = ax.arrow(0, 0, 0, 0, color='blue')

# Text info
text = ax.text(0.05, 0.9, "", transform=ax.transAxes)

ax.set_xlim(-3, 3)
ax.set_ylim(0, 10)
ax.set_xlabel("θ")
ax.set_ylabel("Loss")
ax.set_title("Optimización de θ² con gradiente descendente")
ax.legend()


# ----- 3. ANIMATION FUNCTION -----

def update(frame):
    θ = trajectory[frame]
    grad = grads[frame]

    # Update red point
    point.set_data([θ], [θ**2])

    # Remove old arrow
    global arrow
    arrow.remove()

    # Gradient arrow along x-axis
    scale = 1
    dx = - scale * grad * 0.1   # gradient points right if grad > 0, left if grad < 0
    dy = 0              # no vertical component

    arrow = ax.arrow(θ, θ**2, dx, dy, color='blue', width=0.02)

    # Update text
    text.set_text(f"step={frame}, θ={θ:.3f}, grad={grad:.3f}, loss={θ**2:.3f}")

    return point, arrow, text



anim = FuncAnimation(fig, update, frames=len(trajectory), interval=300)

plt.close()

# ----- 4. DISPLAY IN COLAB -----

HTML(anim.to_jshtml())
```

## Càlcul de la pèrdua

Al exemples manuals hem calcular la pèrdua normalment com `MSE` restant la predicció, elevant al quadrat i traguent la mitjana. No obstant, al igual que passa amb `Scikit-Learn` en `Pythorch` no és necessari fer el càlcul matemàtic i no sempre és tant fàcil com `MSE`. 

Per automatitzar el càlcul de la pèrdua tenim les múltiples funcions de pèrdua: https://docs.pytorch.org/docs/2.14/nn.html#loss-functions 

Seguint l'exemple de regressió lineal, tenim `MSELoss()`.



```python
lr = 0.1

# Step 0 - Initializes parameters "b" and "w" randomly
torch.manual_seed(42)
b = torch.randn(1, requires_grad=True, \
                dtype=torch.float, device=device)
w = torch.randn(1, requires_grad=True, \
                dtype=torch.float, device=device)

# Defines a SGD optimizer to update the parameters
optimizer = torch.optim.SGD([b, w], lr=lr)

# Defines a MSE loss function
loss_fn = torch.nn.MSELoss(reduction='mean')

# Defines number of epochs
n_epochs = 1000

for epoch in range(n_epochs):
    # Step 1 - Computes model's predicted output - forward pass
    yhat = b + w * x_train_tensor
    
    # Step 2 - Computes the loss
    # No more manual loss!
    # error = (yhat - y_train_tensor)
    # loss = (error ** 2).mean()
    loss = loss_fn(yhat, y_train_tensor)

    # Step 3 - Computes gradients for both "b" and "w" parameters
    loss.backward()
    
    # Step 4 - Updates parameters using gradients and
    # the learning rate
    optimizer.step()
    optimizer.zero_grad()
    
print(b, w)
```

## Models

En Pytorch tenim que crear els models a mida. Les solucions anteriors són bones per a problemes menuts d'optimització. Però per a tasques de machine learning que es puguen reproduir i posar en producció cal empaquetar tot en un `Model`. 

Els models són classe heretades de `torch.nn.Module`.

Tot model ha de complir uns requisit i definir unes funcions concretes per a poder ser utilitzat. Aqueste funcions són:

* `__init__(self)`: És el constructor del model i defineix les variables. Al ser el constructor ha de tenir `super().__init__()` per inicialitzar la classe de la que hereta.
* `forward(self,x)`: Realitza el càlcul, fa una predicció durant l'entrenament.

 

```python
class ManualLinearRegression(torch.nn.Module):
    def __init__(self):
        super().__init__()
        # To make "b" and "w" real parameters of the model,
        # we need to wrap them with nn.Parameter
        self.b = torch.nn.Parameter(torch.randn(1,
                                          requires_grad=True, 
                                          dtype=torch.float))
        self.w = torch.nn.Parameter(torch.randn(1, 
                                          requires_grad=True,
                                          dtype=torch.float))
        
    def forward(self, x):
        # Computes the outputs / predictions
        return self.b + self.w * x
```

`Parameter` és un tensor que es considera paràmetre del model. Amb `parameters()` o `state_dict` el podem consultar després.

```python
torch.manual_seed(42)
# Crea una instancia "de prueba" (_dummy_)
# de nuestro modelo ManualLinearRegression
dummy = ManualLinearRegression().to(device)
list(dummy.parameters())
```

```python
dummy.state_dict()
```

Ací tenim la versió amb model de la regressió lineal:

```python
lr = 0.1

# Step 0 - Initializes parameters "b" and "w" randomly
torch.manual_seed(42)
# Now we can create a model and send it at once to the device
model = ManualLinearRegression().to(device)

# Defines a SGD optimizer to update the parameters 
# (now retrieved directly from the model)
optimizer = torch.optim.SGD(model.parameters(), lr=lr)

# Defines a MSE loss function
loss_fn = torch.nn.MSELoss(reduction='mean')

# Defines number of epochs
n_epochs = 1000

for epoch in range(n_epochs):
    model.train() # What is this?!?

    # Step 1 - Computes model's predicted output - forward pass
    # No more manual prediction!
    yhat = model(x_train_tensor)
    
    # Step 2 - Computes the loss
    loss = loss_fn(yhat, y_train_tensor)

    # Step 3 - Computes gradients for both "b" and "w" parameters
    loss.backward()
    
    # Step 4 - Updates parameters using gradients and
    # the learning rate
    optimizer.step()
    optimizer.zero_grad()
    
# We can also inspect its parameters using its state_dict
print(model.state_dict())
```

El podem posar en el bucle anterior o niuat com a part d'un model més gran:

```python
class MyLinearRegression(torch.nn.Module):
    def __init__(self):
        super().__init__()
        # Instead of our custom parameters, we use a Linear model
        # with single input and single output
        self.linear = torch.nn.Linear(1, 1)
                
    def forward(self, x):
        # Now it only takes a call
        self.linear(x)
```

### Models Sequencials

Si volem que l'eixida d'un model siga l'entrada d'un altre podem utilitzar `torch.nn.Sequential` 

```python
torch.manual_seed(42)

model = torch.nn.Sequential(torch.nn.Linear(3, 5), torch.nn.Linear(5, 1)).to(device)

model.state_dict()
```

Analitzem com tenim un model Linear de 3 variables amb 5 eixides. És a dir, múltiple i multivariada (Multi-output). Les eixides són números continus predits pel model. Aquests entren a un altre de 5 entrades i una eixida. 

Per tant, aquestes 5 sortides intermèdies no representen una regressió multivariada final, sinó que actuen com a característiques intermèdies o ocultes (hidden features). La primera capa transforma les 3 dades d'entrada en 5 representacions noves, i la segona capa utilitza aquesta informació per calcular el resultat final.

Això té un nom: `Capes`.

![capes](./capes.png)
