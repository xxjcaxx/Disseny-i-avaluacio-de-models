### Generació de dades

Quan utilitzem Numpy per a tasques de machine learning quasi sempre partim de l'array. Un array de Numpy és igual que una llista de Python, però amb més propietats. 

Podem crear un array de numpy a partir d'una llista. 

````python
my_list = [1,2,3]
import numpy as np
arr = np.array(my_list)
print("Type/Class of this object:",type(arr))
print("Here is the vector\n--------------------\n",arr)
````

De la mateixa manera podem crear matrius. El tipus de dades de tota la matriu serà int64 a no ser que algun element tinga decimals que passarà a ser float64. 

Un array de Numpy no fa distincions en llistes o tuples de Python. 

````python


my_mat = [[1,2,3],[4,5,6],[7,8,9]]
mat = np.array(my_mat)
print("Type/Class of this object:",type(mat))
print("Here is the matrix\n----------\n",mat,"\n----------")
print("Dimension of this matrix: ",mat.ndim,sep='') #ndim gives the dimensison, 2 for a matrix, 1 for a vector
print("Size of this matrix: ", mat.size,sep='') #size gives the total number of elements
print("Shape of this matrix: ", mat.shape,sep='') #shape gives the number of elements along each axes (dimension)
print("Data type of this matrix: ", mat.dtype,sep='') #dtype gives the data type contained in the array

my_mat = [[1.1,2,3],[4,5.2,6],[7,8.3,9]]
mat = np.array(my_mat)
print("Data type of the modified matrix: ", mat.dtype,sep='') #dtype gives the data type contained in the array
print("\n\nEven tuples can be converted to ndarrays...")

b = np.array([(1.5,2,3), (4,5,6)])
print("We write b = np.array([(1.5,2,3), (4,5,6)])")
print("Matrix made from tuples, not lists\n---------------------------------------")
print(b)

````

La capacitat per a generar dades de Numpy és molt versàtil. En mètode `arange` permet crear una serie de números amb l'espai que volem. `linspace` crear la quantitat de números en mig que li passem com a paràmetre. 

````python
print("A series of numbers:",np.arange(5,16)) # A series of numbers from low to high
print("Numbers spaced apart by 2:",np.arange(0,11,2)) # Numbers spaced apart by 2
print("Numbers spaced apart by float:",np.arange(0,11,2.5)) # Numbers spaced apart by 2.5
print("Every 5th number from 50 in reverse order\n",np.arange(50,-1,-5))

print("21 linearly spaced numbers between 1 and 5\n--------------------------------------------")
print(np.linspace(1,5,21))

````

Moltes vegades volem crear arrays de zeros, de uns o aleatoris o buida. Com que està buida, sols ha reservat espai en memòria i pot tenir qualsevol contingut. 

````python
print("Vector of zeroes\n---------------------")
print(np.zeros(5))
print("Matrix of zeroes\n--------------------")
print(np.zeros((3,4))) # Notice Tuples
print("Vector of ones\n---------------------")
print(np.ones(5))
print("Matrix of ones\n---------------------")
print(np.ones((5,2))) # Note matrix dimension specified by Tuples
print("Matrix of 5's\n---------------------")
print(5*np.ones((3,5)))
print("Empty matrix\n-------------\n", np.empty((3,5)))
````

````python
mat1 = np.eye(4) 
print("Identity matrix of dimension", mat1.shape)
print(mat1)
````

Generar dades aleatòries és una pràctica molt comú per a fer proves de models o per a inicialitzar models. Numpy permet crear números aleatoris, aleatoris amb una distribució normal o integers amb uns rangs i forma determinats. 

````python
print("Random number generation (from Uniform distribution)")
print(np.random.rand(2,3)) # 2 by 3 matrix with random numbers ranging from 0 to 1, Note no Tuple is necessary 
print("Numbers from Normal distribution with zero mean and standard deviation 1 i.e. standard normal")
print(np.random.randn(4,3))
print("Random integer vector:",np.random.randint(1,100,10)) #randint (low, high, # of samples to be drawn)
print ("\nRandom integer matrix")
print(np.random.randint(1,100,(4,4))) #randint (low, high, # of samples to be drawn in a tuple to form a matrix)
print("\n20 samples drawn from a dice throw:",np.random.randint(1,7,20)) # 20 samples drawn from a dice throw
````

### Reshape

Capítol a part mereix la funció `reshape`, ja que sovint els datasets tenen una forma no desitjada. 

A l'exemple tenim una matriu de una dimensió i 30 elements. La funció `reshape` accepta les noves dimensions que volem i crea una matriu amb eixa forma. En aquest cas una matri tridimensional amb 3 files de 5 columnes i una profunditat de 2. 

````python
from numpy.random import randint as ri
a = ri(1,100,30)
b = a.reshape(2,3,5)
c = a.reshape(6,5)
print ("Shape of a:", a.shape)
print ("Shape of b:", b.shape)
print ("Shape of c:", c.shape)
print("\na looks like\n",'-'*20,"\n",a,"\n",'-'*20)
print("\nb looks like\n",'-'*20,"\n",b,"\n",'-'*20)
print("\nc looks like\n",'-'*20,"\n",c,"\n",'-'*20)
````

Veiem com es comporta `sort` amb matrius de distintes formes. 

En el primer cas simplement ordena l'array unidimensional. El comandament sort permet ordenar cadascuna de les files o les columnes amb `axis=0`. Observa que ordena segons el primer element de cada columna original.  

````python

A = ri(1,100,10) # Vector of random integers
print("\nVector of random integers\n",'-'*50,"\n",A)
print("\nHere is the sorted vector\n",'-'*50,"\n",np.sort(A, kind='mergesort'))

M = ri(1,100,25).reshape(5,5) # Matrix of random integers
print("\n5x5 Matrix of random integers\n",'-'*50,"\n",M)
print("\nHere is the sorted matrix along each row\n",'-'*50,"\n",np.sort(M, kind='mergesort')) # Default axis =1
print("\nHere is the sorted matrix along each column\n",'-'*50,"\n",np.sort(M, axis=0, kind='mergesort'))
````

### Índex dels arrays

Els arrays de Numpy són més flexibles per a accedir a posicions aleatories que les llistes de Python. A continuació veiem l'ús de eś `:` per a expressar rangs (slicing), el -1 per a indicar l'última posició, el tercer número que indica els passos i l'ús de llistes d'index. 

````python
arr = np.arange(0,11)
print("Array:",arr)
print("Element at 7th index is:", arr[7])
print("Elements from 3rd to 5th index are:", arr[3:6])
print("Elements up to 4th index are:", arr[:4])
print("Elements from last backwards are:", arr[-1::-1]) # de -1 (última posició) fins a la primera posició, amb pas -1 (decrement)
print("3 Elements from last backwards are:", arr[-1:-6:-2]) # de -1 (última posició) fins a la sisena posició, amb pas -2 (decrement)

arr = np.arange(0,21,2)
print("New array:",arr)
print("Elements at 2nd, 4th, and 9th index are:", arr[[2,4,9]]) # Pass a list as a index to subset
````

En els cas de les matrius resulta especialment còmode accedir a les posicions sense tenir que posar tots els `[]`. 

Amés, extreure una columna és tant fàcil com una fila amb `:` Ahí indiquem que de totes les files volem determinades columnes. 

Amb aquestes possibilitats es poden fer moltes combinacions. 

````python


mat = np.array(ri(10,100,15)).reshape(3,5)
print("Matrix of random 2-digit numbers\n--------------------------------\n",mat)

print("\nDouble bracket indexing\n------------------------")
print("Element in row index 1 and column index 2:", mat[1][2])

print("\nSingle bracket with comma indexing\n----------------------------------")
print("Element in row index 1 and column index 2:", mat[1,2])
print("\nRow or column extract\n----------------------")

print("Entire row at index 2:", mat[2])
print("Entire column at index 3:", mat[:,3])

print("\nSubsetting sub-matrices\n--------------------------")
print("Matrix with row indices 1 and 2 and column indices 3 and 4\n", mat[1:3,3:5])
print("Matrix with row indices 0 and 1 and column indices 1 and 3\n", mat[0:2,[1,3]])

````

També es pot buscar als arrays o matriu per criteris de búsqueda. Observa l'exemple i després la seua descomposició:

````python
mat = np.array(ri(10,100,15)).reshape(3,5)
print("Matrix of random 2-digit numbers\n--------------------------------\n",mat)
print ("Elements greater than 50\n", mat[mat>50])

print("Descomposant la búsqueda", mat>50)
# Aquesta técnica de subsetting es coneix com a "boolean indexing" i és molt útil per filtrar elements d'un array basant-se en una condició.
# A Pandas també es pot fer servir aquesta tècnica per filtrar files d'un DataFrame basant-se en condicions aplicades a les columnes.
````

Les matrius de les quals s'extrau un subset no són copies, si modifiquem la submatriu l'original manté la referència a les dades i muta també. 

Si volem fer una copia podem passar les dades a `np.array` de nou. 


````python
mat = np.array([[11,12,13],[21,22,23],[31,32,33]])
print("Original matrix")
print(mat)
mat_slice = mat[:2,:2]
print ("\nSliced matrix")
print(mat_slice)
print ("\nChange the sliced matrix")
mat_slice[0,0] = 1000
print (mat_slice)
print("\nBut the original matrix? WHOA! It got changed too!")
print(mat)

# Little different way to create a copy of the slixed matrix
print ("\nDoing it again little differently now...\n")
mat = np.array([[11,12,13],[21,22,23],[31,32,33]])
print("Original matrix")
print(mat)
mat_slice = np.array(mat[:2,:2]) # Notice the np.array command to create a new array not just slicing
print ("\nSliced matrix")
print(mat_slice)
print ("\nChange the sliced matrix")
mat_slice[0,0] = 1000
print (mat_slice)
print("\nBut the original matrix? NO CHANGE this time:)")
print(mat)
````

### Operacions vectorials 

Una de les característiques més potents del Numpy són les operacions vectorials. En compte de fer "fors" per fer operacions matemàtiques amb tots els elements, Numpy treballa de forma natural amb vectors.  

````python


mat1 = np.array(ri(1,10,9)).reshape(3,3)
mat2 = np.array(ri(1,10,9)).reshape(3,3)
print("\n1st Matrix of random single-digit numbers\n----------------------------------------\n",mat1)
print("\n2nd Matrix of random single-digit numbers\n----------------------------------------\n",mat2)

print("\nAddition\n------------------\n", mat1+mat2)
print("\nMultiplication\n------------------\n", mat1*mat2)
print("\nDivision\n------------------\n", mat1/mat2)
print("\nLineaer combination: 3*A - 2*B\n-----------------------------\n", 3*mat1-2*mat2)

print("\nAddition of a scalar (100)\n-------------------------\n", 100+mat1)

print("\nExponentiation, matrix cubed here\n----------------------------------------\n", mat1**3)
print("\nExponentiation, sq-root using pow function\n-------------------------------------------\n",pow(mat1,0.5))

````

Entre les operacions vectorials, el `broadcasting` permet fer operacions amb matrius de diferents dimensions. A continuació tenim un exemple de boradcast en files, en columnes i en les dos direccions. 

````python
start = np.zeros((4,3))
print(start)

# create a rank 1 ndarray with 3 values
add_rows = np.array([1, 0, 2])
print(add_rows)

y = start + add_rows  # add to each row of 'start' using broadcasting
print(y)


# create an ndarray which is 4 x 1 to broadcast across columns
add_cols = np.array([[0,1,2,3]])
print(add_cols)
add_cols = add_cols.T # transpose to make it 4 x 1 with .T
print(add_cols)

y = start + add_cols  # add to each column of 'start' using broadcasting
print(y)

# this will just broadcast in both dimensions
add_scalar = np.array([100])  
print(start+add_scalar)

````

Ací tenim algunes operacions disponibles a Numpy que també es poden aplicar vectorialment a matrius. 

````python
mat1 = np.array(ri(1,10,9)).reshape(3,3)
mat2 = np.array(ri(1,10,9)).reshape(3,3)
print("\n1st Matrix of random single-digit numbers\n----------------------------------------\n",mat1)
print("\n2nd Matrix of random single-digit numbers\n----------------------------------------\n",mat2)

print("\nSq-root of 1st matrix using np\n------------------\n", np.sqrt(mat1))
print("\nExponential power of 1st matrix using np\n",'-'*50,"\n", np.exp(mat1))
print("\n10-base logarithm on 1st matrix using np\n",'-'*50,"\n", np.log10(mat1))
# fmod és el mòdul de la divisió, és a dir, el residu de la divisió entre els elements corresponents de les dues matrius. La funció np.fmod retorna aquest residu element a element.
print("\nModulo reminder using np\n",'-'*50,"\n", np.fmod(mat1,mat2))
````

El següent exemple il·lustra com es poden combinar aquestes operacions. El que fa és crear un array de 1001 elements de 0 a $12*\pi$ i mostra una gràfica d'una ona amb un decay exponencial. Mostra com canvia el signe d'una matriu, fa una divisió vectorial o calcula el seno. 

`y` és un altre array calculat vectorialment abans de fer la gràfica

````python
import matplotlib.pyplot as plt

print("\nCombination of functions by shwoing exponetial decay of a sine wave\n",'-'*70)
A = np.linspace(0,12*np.pi,1001)
plt.scatter(x=A,y=100*np.exp(-A/10)*(np.sin(A)))
plt.title("Exponential decay of sine wave: exp(-x)*sin(x)")
plt.show()
````

Amb `axis=0` indiquem que volem fer les operacions en les columnes i `axis=1` en les files. 

Les operacions bàsiques de l'estadística com la `mitjana`, la `desviació típica` o el `percentils` també són aplicables a matrius, files i columnes. 

````python


mat1 = np.array(ri(1,10,9)).reshape(3,3)
mat2 = np.array(ri(1,10,9)).reshape(3,3)
print("\n1st Matrix of random single-digit numbers\n","-"*50,"\n",mat1)
print("\n2nd Matrix of random single-digit numbers\n","-"*50,"\n",mat2)

print("\nSum of all numbers in 1st matrix\n","-"*50,"\n",np.sum(mat1))
print("\nSum of all numbers in columns of 1st matrix\n","-"*50,"\n",np.sum(mat1,axis=0))
print("\nSum of all numbers in rows of 1st matrix\n","-"*50,"\n",np.sum(mat1,axis=1))
print("\nProduct of all numbers in rows of 1st matrix\n","-"*50,"\n",np.prod(mat1,axis=1))
print("\nProduct of all numbers in columns of 2nd matrix\n","-"*50,"\n",np.prod(mat2,axis=0))
print("\nMean of all numbers in 1st matrix\n","-"*50,"\n",np.mean(mat1))
print("\nStandard deviation of all numbers in 1st matrix\n","-"*50,"\n",np.std(mat1))

mat1 = np.array(ri(1,100,9)).reshape(3,3)
print("\nModified matrix of random numbers from 1 to 99\n","-"*50,"\n",mat1)
print("\nStandard deviation of all numbers in the modified matrix, a larger number\n","-"*80,"\n",np.std(mat1))
print("\nVariance of all numbers in the modified matrix, a larger number\n","-"*80,"\n",np.var(mat1))
print("\nMedian of all numbers in the modified matrix\n","-"*60,"\n",np.median(mat1))

mat2 = np.array(ri(1,100,50)).reshape(10,5)
print("\nModified matrix of 50 random numbers from 1 to 99\n","-"*50,"\n",mat2)
print("\nStandard deviation along the columns in the modified matrix\n","-"*60,"\n",np.std(mat2,axis=0))

mat1 = np.array(ri(1,100,20)).reshape(4,5)
print("\nModified matrix of random numbers from 1 to 99\n","-"*50,"\n",mat1)
print("\nFlattened and sorted matrix (as vector)\n","-"*50,"\n",np.sort(mat1.reshape(1,20)))
print("\n50th percentile of all numbers in the modified matrix\n","-"*60,"\n",np.percentile(mat1,50))
print("\n90th percentile of all numbers in the modified matrix\n","-"*60,"\n",np.percentile(mat1,90))

````

### Descomposició de Valors Singulars

`np.linalg.svd` realitza la **Descomposició en Valors Singulars** (SVD, per les seves sigles en anglès: *Singular Value Decomposition*). És un dels teoremes i eines més importants de l'àlgebra lineal i de la ciència de dades.

En poques paraules, l'SVD agafa qualsevol matriu $A$ (ja sigui quadrada o rectangular) i la descompon en el producte de **tres matrius especials**:

$$A = U \cdot S \cdot V^T$$

Si tens una matriu original $A$ de dimensions $m \times n$, `np.linalg.svd(A)` et retorna tres elements:

1. **`U` (Matriu ortogonal esquerra):** Les seves columnes són els anomenats *vectors singulars esquerra*. Té dimensions $m \times m$.
2. **`s` (Valors singulars):** Un array 1D que conté els anomenats *valors singulars* ordenats de major a menor. Aquests valors representen "la importància" o "la força" de cada dimensió o component de la matriu.
3. **`Vh` ($V^T$, Matriu ortogonal dreta):** Les files són els *vectors singulars dreta* (atenció: NumPy ja et retorna aquesta matriu directament **transposada**). Té dimensions $n \times n$.

> Si vols utilitzar directament els valors singulars per reconstruir la matriu o fer càlculs, NumPy sol retornar `s` com un array simple. Per fer multiplicacions de matrius, sovint es converteix `s` en una matriu diagonal amb `np.diag(s)`.

L'SVD s'utilitza a tot arreu perquè permet simplificar, comprimir i analitzar dades complexes:

1. **Reducció de dimensionalitat (PCA / Principal Component Analysis):** El PCA es pot calcular directament utilitzant l'SVD de la matriu de dades. Permet reduir moltes variables a unes poques conservant la màxima informació possible.
2. **Compressió d'imatges:** Com que els valors singulars (`s`) estan ordenats del més important al menys important, pots descartar els valors més petits (fent una aproximació de baix rang) per reduir dràsticament la mida d'una imatge o d'una matriu sense perdre gairebé qualitat.
3. **Sistemes de recomanació (Collaborative Filtering):** Es fa servir famosa-ment per predir valoracions d'usuaris (com feia Netflix per recomanar pel·lícules) descomponent la matriu d'usuaris-productes.
4. **Càlcul de la inversa generalitzada (Pseudoinversa):** Permet resoldre sistemes d'equacions lineals fins i tot quan la matriu no té inversa clau (per exemple, si no és quadrada).

````python
A = np.random.randint(1,10,9).reshape(3,3)
print("Original matrix\n",A)
print("\n")
u,s,v=np.linalg.svd(A,compute_uv=1,full_matrices=True)
print("u:",u)
print("\n")
print("Singular values, s:",s)
print("\n")
print("v:",v)
print("\n")
print ("Reconstruction of A, u*s*v\n",np.dot(u,np.dot(np.diag(s),v)))
````

### Descomposició QR

`np.linalg.qr` realitza la **Descomposició QR** d'una matriu. És una altra operació fonamental d'àlgebra lineal que agafa qualsevol matriu $A$ (habitualment de dimensions $m \times n$) i la separa en el producte de **dues matrius especials**:

$$A = Q \cdot R$$

Si apliquem la descomposició a una matriu $A$, `np.linalg.qr(A)` retorna dos elements:

1. **`Q` (Matriu Ortogonal / Orthonormal):** Una matriu on les seves columnes són vectors unitaris i ortogonals entre si (formen un angle de 90 graus). Tècnicament, compleix que $Q^T \cdot Q = I$ (la identitat).
2. **`R` (Matriu Triangular Superior):** Una matriu on tots els elements situats *per sota* de la diagonal principal són zero ($R_{i,j} = 0$ si $i > j$).

La descomposició QR s'utilitza principalment per resoldre problemes numèrics de manera molt estable i eficient:

1. **Resolució de sistemes d'equacions lineals ($Ax = b$):** En lloc de calcular la inversa d'una matriu (que pot ser inestable numèricament), utilitzar la factorització QR simplifica el càlcul a resoldre $R x = Q^T b$, la qual cosa es fa molt ràpidament per substitució cap enrere gràcies a la forma triangular de $R$.
2. **Càlcul de Mínims Quadrats:** És el mètode que hi ha darrere de molts algorismes de regressió per trobar la línia o pla que millor s'ajusta a un conjunt de dades sorolloses sense patir problemes d'arrodoniment numèric.
3. **Càlcul d'Autovalors (Algorisme QR):** S'utilitza de forma iterativa per trobar els autovalors (*eigenvalues*) d'una matriu de manera molt eficient.```

````python
A = np.random.randint(1,10,9).reshape(3,3)
print("Original matrix\n",A)
print("\n")
q,r=np.linalg.qr(A)
print("Q:",q)
print("\n")
print("R:",r)
print("\n")
print ("Reconstruction of A, Q*R\n",np.dot(q,r))
````

### Eigenvalues

`np.linalg.eig` calcula els **autovalors** (*eigenvalues*) i els **autovectors** (*eigenvectors*) d'una matriu quadrada.

En àlgebra lineal, un autovector d'una matriu és un vector que, quan se li aplica la transformació de la matriu, **no canvia de direcció**, sinó només de magnitud (s'escala). El factor d'escalat és precisament l'autovalor. Matemàticament es compleix que:

$$A \cdot v = \lambda \cdot v$$

On:

* $A$ és la matriu.
* $v$ és l'autovector.
* $\lambda$ (lambda) és l'autovalor associat.

Quan executem `np.linalg.eig(A)` sobre una matriu quadrada de dimensions $n \times n$, retorna una tupla amb dos elements:

1. **`w` (Autovalors):** Un array 1D de longitud $n$ que conté els valors escalars ($\lambda$).
2. **`v` (Autovectors):** Una matriu de dimensions $n \times n$ on **cada columna** correspon a un autovector de la matriu (la columna `v[:, i]` correspon a l'autovalor `w[i]`).

L'anàlisi d'autovalors i autovectors és el cor de moltes tècniques avançades:

1. **PCA (Principal Component Analysis / Anàlisi de Components Principals):** És l'ús més famós en ML. Per reduir dimensions d'un conjunt de dades, es calcula la matriu de covariància de les dades i s'obtenen els seus autovalors i autovectors. Els autovectors amb els autovalors més grans marquen les "direccions" on hi ha més variació (les components principals).
2. **PageRank de Google:** L'algorisme original de cerca de Google utilitza els autovectors d'una matriu de transició gigantesca per calcular la importància de cada pàgina web a internet.
3. **Sistemes dinàmics i Xarxes Neuronals:** Permet estudiar l'estabilitat d'est sistemes, analitzant com evoluciona una informació o un senyal al llarg del temps quan es multipliquen iterativament per una matriu de pesos.

````python


A = np.random.randn(9).reshape(3,3)
print("Original matrix\n",A)
print("\n")
w,v=np.linalg.eig(A)
print("Eigenvalues:\n",w)
print("\n")
print("Eigenvectors:\n",v)

````

### Resoldre sistemes d'ecuacions

1. $3x + y = 9$
2. $x + 2y = 8$

Això en format matricial es pot escriure com $\begin{pmatrix} 3 & 1 \\ 1 & 2 \end{pmatrix} \cdot \begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} 9 \\ 8 \end{pmatrix}$.


````python
import numpy as np

# Matriu de coeficients (A)
A = np.array([[3, 1],
              [1, 2]])

# Vector de resultats (b)
b = np.array([9, 8])

# Resolem el sistema Ax = b
x = np.linalg.solve(A, b)

print("Solució (x, y):", x)
# Sortida: [2.  3.]  (significa que x = 2 i y = 3)
````

### Regressió lineal

Numpy té la seua propia manera de fer regressió lineal sense necessitat de Scikit-Learn. No fa un model per posar en producció a la manera de `Scikit-Learn`, però per a fer estadística descriptiva o EDA pot ser molt àgil. 

Hi ha dos maneres de calcular la regressió lineal, amb `np.polyfit` i amb `np.lstsq`. 

Amb una sola variable independent `polyfit` és molt còmode:

````python
import numpy as np

# Dades d'exemple
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Calculem la regressió lineal (grau 1)
# Retorna [pendent (m), ordenada al origen (c)]
m, c = np.polyfit(x, y, 1)

print(f"Pendent (m): {m}, Pendent al tall (c): {c}")
````

Si tenim múltiples variables ja necessitem `lstsq`, que calcula els mínims quadrats. Aquesta funció necessita una matriu amb les variables i el terme independent `c`. 

$$y = m \cdot x + c$$

Ha de resoldre un sistema d'equacions per trobar dos valors diferents ($m$ i $c$) mitjançant una multiplicació de matrius. Necessita una columna de 1s per a que retorne sempre `c`. 


````python


x = np.arange(1,11,1)
y = 2*x+np.random.randn(10)-1
print(x)
print(y)
A = np.vstack([x, np.ones(len(x))]).T 
print(A)

m, c = np.linalg.lstsq(A, y)[0]
print ("Coefficient:"+str(m)+ "\n"+"intercept:"+str(c))

# Plot the fitteed line
import matplotlib.pyplot as plt
%matplotlib inline
plt.plot(x, y, 'o', label='Original data', markersize=10)
plt.plot(x, m*x + c, 'r', label='Fitted line')
plt.legend()
plt.show()


````

El següent exemple posa la columna de 1s davant i per això retorna en primer lloc el `intercept` de la recta. 

````python
import numpy as np

# Matriu de característiques (afegint una columna de uns per al terme independent)
X = np.array([[1, 1], [1, 2], [1, 3], [1, 4]])
y = np.array([2, 4, 5, 4])

# Resolem per mínims quadrats
pesos, residuals, rank, s = np.linalg.lstsq(X, y, rcond=None)
print("Pesos òptims:", pesos)
print("Equació de la recta ajustada: y = {:.2f} + {:.2f}x".format(pesos[0], pesos[1]))
````

### Inversa i pseudoinversa

`np.linalg.inv` vs `np.linalg.pinv` (Inversa vs Pseudoinversa)

Totes dues funcions s'utilitzen per trobar la "inversa" d'una matriu, és a dir, una matriu que, en multiplicar-la per l'original, ens dona la matriu identitat. 

La pinv no falla en matrius no quadrades i sempre dona una solució aproximada sense falla. 

Són utilitzades per a calcular la regressió lineal. 


````python
A = 0.1*np.random.randint(1,20,16).reshape(4,4)
print ("A:\n",A)
print ("Inverse of A:\n",np.linalg.inv(A))
````

````python


A = 0.1*np.random.randint(1,20,15).reshape(5,3)
print ("A:\n",A)
print ("Pseudo-inverse of A:\n",np.linalg.pinv(A))
print ("Matrix product of A and pseudo inverse:\n", np.dot(np.linalg.pinv(A),A))

````
